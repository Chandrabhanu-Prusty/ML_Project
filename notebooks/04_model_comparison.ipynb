# Holdout model comparison

Read frozen holdout metrics after validation chooses the model and threshold. Ranking by holdout AP is descriptive and does not change the selected model.

Run the training CLI before executing this notebook.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import Image, display

root = next(path for path in (Path.cwd(), *Path.cwd().parents)
            if (path / 'configs/full.json').is_file())
sys.path.insert(0, str(root))
from src.utils import read_json

pointer = root / 'models/latest.json'
if not pointer.is_file():
    raise FileNotFoundError('Run python main.py --mode train from the project root first.')
run = root / 'reports/runs' / read_json(pointer)['run_id']
metadata = read_json(run / 'metadata.json')
print(metadata['profile'], metadata['run_id'])
```

```python
results = pd.DataFrame.from_dict(metadata['models'], orient='index')
display(results[[
    'cv_average_precision', 'selection_average_precision', 'average_precision',
    'roc_auc', 'accuracy', 'precision', 'recall', 'f1', 'false_positives',
    'false_negatives', 'threshold', 'training_seconds', 'inference_microseconds_per_row'
]].sort_values('average_precision', ascending=False))
print('Validation selected:', metadata['best_model'])
for name in ['comparison', 'curves', 'confusion_matrices']:
    display(Image(filename=str(run / 'figures' / f'{name}.png')))
```
